```python
import os
import shutil
import random

# Set the path to the directory containing the original dataset
data_dir = 'E:\cricket\\New folder\Dataset\\'

# Set the path to the directory where you want to create the train and validation directories
output_dir = 'E:\cricket\\New folder\split_dataset\\'

# Set the percentage of data to use for validation
validation_split = 0.3

# Create the train and validation directories
train_dir = os.path.join(output_dir, 'train')
valid_dir = os.path.join(output_dir, 'valid')
os.makedirs(train_dir, exist_ok=True)
os.makedirs(valid_dir, exist_ok=True)

# Loop over the subdirectories in the original dataset directory
for subdir in os.listdir(data_dir):
    subdir_path = os.path.join(data_dir, subdir)
    # Create subdirectories in the train and validation directories
    train_subdir = os.path.join(train_dir, subdir)
    valid_subdir = os.path.join(valid_dir, subdir)
    os.makedirs(train_subdir, exist_ok=True)
    os.makedirs(valid_subdir, exist_ok=True)
    # Loop over the files in the subdirectory
    files = os.listdir(subdir_path)
    random.shuffle(files)  # shuffle the files randomly
    num_valid1 = int(len(files)*0.3)
    num_valid2 = int(len(files)*0.5)
    print(num_valid1,num_valid2)
    valid_files = files[:num_valid1]
    train_files = files[num_valid1:]
    # Move the files into the train and validation subdirectories
    for file in train_files:
        src_path = os.path.join(subdir_path, file)
        dest_path = os.path.join(train_subdir, file)
        shutil.copy(src_path, dest_path)
    for file in valid_files:
        src_path = os.path.join(subdir_path, file)
        dest_path = os.path.join(valid_subdir, file)
        shutil.copy(src_path, dest_path)
```
